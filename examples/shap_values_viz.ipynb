```python
import shap
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from riskpilot.evaluation import BinaryPerformanceEvaluator

# 1. Carregar dataset
X, y = load_breast_cancer(return_X_y=True, as_frame=True)
df = X.copy()
df["target"] = y
df["id"] = range(len(df))
df["date"] = pd.date_range("2020-01-01", periods=len(df), freq="D")

# 2. Split
df_train, df_test = train_test_split(df, test_size=0.3, random_state=42)
```

```python
# 3. Treinar modelo
features = df_train.drop(columns=["target", "id", "date"]).columns
model = LogisticRegression(max_iter=1000).fit(df_train[features], df_train["target"])

# 4. Instanciar o evaluator
evaluator = BinaryPerformanceEvaluator(
    model=model,
    df_train=df_train,
    df_test=df_test,
    target_col="target",
    id_cols=["id"],
    date_col="date",
)
```

```python
# 5. Plotar visualizações SHAP de treino e teste
figs = evaluator.plot_shap_beeswarm(
    #splits=["train", "test"],
    #plot_type="bar",             # ou "layered", "beeswarm", "trend", etc.
    #reference_split="train",
    #summary=True,
    #save=False                   # ou True para salvar
)

# 6. Visualizar
#figs["figures"].show()
```
